```python
oni = pd.read_csv("../oni_mensual_1981_2026.csv", parse_dates=["time"])
rain = pd.read_csv("../precipitacion_ensenada_1981_2026.csv", parse_dates=["time"])
```

```python
#Set_index
oni = oni.set_index('time')
rain = rain.set_index('time')
```

## Concatenar (Unir los datos en una tabla)

```python
#Concatenar 
analysis = pd.concat(
    [
        ssta_series.rename('ssta_degC'),
        oni['oni_degC'],
        rain['precip_mm']
    ],
    axis=1
)

analysis.head()
```

## Hacer una función 

```python
def regional_mean(da, lat_range, lon_range):
    """ Calcula la media espacial de un DataArray en una región rectangular.
    
    Parameters
    ----------
    da : xarray.DataArray
         campo con dimensiones lat y lon
    lat_range : tuple
         (lat_min, lat_max)
    lon_range : tuple
         (lon_min, lat_max)

    Returns
    ---------
    xarray.DatArray
        Serie resultante después de promediar lat y lon
    """
    region = da.sel(lat=slice(*lat_range), lon=slice(*lon_range))

    return(region.mean(dim=('lat', 'lon'), skipna=True))
    
```

```python
baja_series = regional_mean(ds.ssta, (28,33), (-120, -115))
baja_series
```
